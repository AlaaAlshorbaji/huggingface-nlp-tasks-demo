What is Text Classification?
Text classification is the task of assigning a label or category to a given piece of text.
It’s one of the most common tasks in NLP and has many real-world applications.

-----------------------------------------------------------------

Major Types of Text Classification

Detect whether a text expresses positive, negative, or neutral emotion.

2. Spam Detection

# Task 2: Question Answering

What is Question Answering?

Question Answering (QA) is the task where an AI system is given:

a context (a paragraph or document), and

a question related to that context

The model must find and return the correct answer from the context.



---------------------------------------------

Types of Question Answering

We’ll focus now on Extractive Question Answering, where the answer is always part of the input text.

Examples of other types (for later):

Abstractive QA (closed-book) – generates the answer from learned knowledge

Multiple choice QA – picks from given options

Open-domain QA – searches a large corpus

## 1.Extractive Question Answering

 What model will we use?

We'll use:
deepset/roberta-base-squad2
A powerful RoBERTa model fine-tuned on the SQuAD 2.0 dataset for QA.



```python
from transformers import pipeline

# Load QA pipeline with a QA model
qa_pipeline = pipeline("question-answering", model="deepset/roberta-base-squad2")

# Provide context and question
context = """
Elon Musk is the CEO of Tesla and SpaceX. He was born in South Africa
and later moved to the United States to pursue business and technology ventures.
"""

question = "Where was Elon Musk born?"

# Get the answer
result = qa_pipeline(question=question, context=context)

# Print the answer
print(result)
```

## Closed-Book Question Answering

What is Closed-Book QA?

In Extractive QA, the model finds answers inside a given paragraph.

In Closed-Book QA, there is no context paragraph — the model answers based only on its internal knowledge learned during training.

You just ask a question like:

“Who wrote the Harry Potter books?”

And the model generates the answer directly, like:

“J.K. Rowling”

🤖 What kind of models do Closed-Book QA?
Closed-book QA uses generative models trained to answer questions directly, such as:

t5-base or t5-large

google/flan-t5-base

facebook/bart-large

bigscience/T0

google/palm or llama models (in advanced setups)

We’ll use a fast and good one:
google/flan-t5-base



```python
from transformers import pipeline

# Load a text2text-generation model (T5)
qa_pipeline = pipeline("text2text-generation", model="google/flan-t5-base")

# Ask a question — no context provided
question = "Who discovered gravity?"

# The model will try to generate the answer
result = qa_pipeline(question)

print(result)
```

How does this work?

The model has learned answers to many factual questions from its pretraining on massive datasets like:

Wikipedia

Books

Web articles

So it remembers common knowledge and can generate answers from memory.

❗ Limitations of Closed-Book QA
Sometimes it hallucinates (gives wrong answers confidently)

It cannot answer domain-specific questions unless trained on them

✅ Summary

Closed-Book QA = no context, just a question → model answers

Based on internal knowledge only

Uses generative models like T5, BART, FLAN-T5



## Multiple Choice Question Answering

What is Multiple Choice QA?

In this QA format, the model is given:

A question

A list of possible answers

Sometimes a context paragraph

Its job is to choose the correct option from the given choices.

📍 Real-life examples

Exams (e.g. TOEFL, GRE, SAT)

AI assistants narrowing down answers

Quiz bots and e-learning tools

🤖 What models support it?

You need models fine-tuned for multiple-choice tasks like:

bert-base-uncased (with custom head)

roberta-base (customized for multiple choice)

valhalla/distilbart-mnli-12-3 → works out-of-the-box using zero-shot-classification pipeline
(It uses NLI: "Does this sentence support this label?")

We'll use the zero-shot approach — simple, clean, no training.

```python
from transformers import pipeline

# Load a zero-shot classification pipeline
multi_choice = pipeline("zero-shot-classification", model="valhalla/distilbart-mnli-12-3")

# Question
question = "What is the capital of France?"

# Choices
choices = ["Berlin", "Paris", "Madrid", "London"]

# Run model
result = multi_choice(question, candidate_labels=choices)

print(result)
```

How does this work?

This is a trick using Natural Language Inference (NLI).

The model checks: “Does this sentence imply the label X?”

It ranks all candidate labels and chooses the best match.

So we treat the question as a "hypothesis" and labels as "premises."

# Task 3: Text Summarization

 What is Summarization?

Summarization is the task of taking a long piece of text and generating a shorter version that keeps the most important information.

There are two main types:

1.Extractive summarization

Picks important sentences from the original text
(like cutting and pasting the best parts)


2.Abstractive summarization

Generates new sentences to paraphrase and compress the original
(like how a human would rewrite the key points)

We’ll focus on abstractive summarization, which is the default in Hugging Face pipeline.

📍 Real-world examples

Summarizing news articles

Creating executive summaries of long reports

Reducing long customer reviews

Helping students condense textbooks

🤖 What models are used?

The best models for summarization include:

facebook/bart-large-cnn ✅

google/pegasus-xsum

t5-base, t5-large

google/flan-t5

We’ll use the most popular:

facebook/bart-large-cnn

It’s trained on CNN/DailyMail news articles to generate high-quality summaries.

```python
from transformers import pipeline

# Load summarization pipeline
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

#This loads a BART model fine-tuned for summarizing news articles.
#The model reads a long text and tries to rewrite the main idea in fewer words.



# Long article or paragraph
text = """
The Eiffel Tower is one of the most recognizable landmarks in the world.
Located in Paris, France, it was constructed for the 1889 World's Fair and stands
at a height of 300 meters. It was initially criticized by some of France's leading
artists and intellectuals for its design, but has since become a global cultural icon
of France and one of the most visited monuments in the world.
"""

# Run summarization
summary = summarizer(text, max_length=50, min_length=25, do_sample=False)


#max_length: the maximum length of the generated summary

#min_length: the minimum length of the summary


#do_sample=False: makes output deterministic (not random)



# Print summary
print(summary[0]['summary_text'])


#The model returns a list with one item (the summary).
#This extracts the text of the summary itself.
```

# Task 5: Paraphrasing

What is Paraphrasing?

Paraphrasing is the task of rewriting a sentence using different words, while keeping the same meaning.

Example:

Original: “She is very smart.”
Paraphrased: “She has a brilliant mind.”

----------------------------------------------------------------

t’s useful for:

Rewriting content

Avoiding repetition

Generating alternative answers

Data augmentation

 Which models are good for paraphrasing?
We’ll use:

T5-based models, like Vamsi/T5_Paraphrase_Paws

ramsrigouthamg/t5_paraphraser

You can also use flan-t5 with prompts

We’ll use a clean and effective one:
✅ Vamsi/T5_Paraphrase_Paws

It’s trained on the PAWS dataset (Paraphrase Adversaries from Word Scrambling).

```python
from transformers import pipeline

# Load the model
paraphraser = pipeline("text2text-generation", model="google/flan-t5-large")

# A longer and more contextual input for better paraphrasing
text = "Artificial intelligence is transforming global industries. It helps companies automate processes and make smarter decisions faster."

# Updated prompt for more creativity
prompt = f"Come up with 5 different ways to say this sentence with the same meaning but different words: {text}"

# Generate paraphrases using pure sampling (beam search OFF)
results = paraphraser(
    prompt,
    max_length=100,
    num_return_sequences=5,
    do_sample=True,
    temperature=1.9,
    top_k=50,
    top_p=0.92,
    num_beams=1      # ✅ This disables the default beam search
)

# Show results
print("Original:", text)
print("\nParaphrases:")
for i, r in enumerate(results):
    print(f"{i+1}. {r['generated_text']}")
```

Loads the FLAN-T5 Large model from Hugging Face.

"text2text-generation" is used for models trained to convert one piece of text into another — like summarization, translation, or paraphrasing.

This model is instruction-tuned — it understands prompts like "Paraphrase this", "Translate this", etc.

 Creating a clear, effective prompt

prompt = f"Come up with 5 different ways to say this sentence with the same meaning but different words: {text}"

This prompt is very important.

FLAN-T5 responds better when you give it a clear instruction.


The prompt tells the model to paraphrase, not summarize or translate.



Let’s break this down:

Argument	What It Means

prompt	The instruction string you created earlier.

max_length=100	Limits the total length of the generated text.

num_return_sequences=5	You want 5 different paraphrases.

do_sample=True	Enables random sampling instead of greedy
 or beam search decoding.

temperature=1.9	Controls randomness: higher = more creative. Default is 1.0.

top_k=50	Limits sampling to top 50 likely next tokens. Helps creativity.

top_p=0.92	Nucleus sampling: only sample from top 92% of cumulative probability.

num_beams=1	Disables beam search (which is deterministic and repetitive).


→ These values combined give you creative, unique, and high-quality paraphrases.



Prints the original sentence.

Loops through the generated results and prints each paraphrased version with a number.

# Task 7: Sentence Similarity

What Is Sentence Similarity?

This task answers the question:
“Do two sentences mean the same thing?”

It’s commonly used in:

✅ Duplicate detection (e.g. “Did the user ask this already?”)

✅ Semantic search (e.g. “Find similar questions/articles”)

✅ Automatic grading (e.g. “Did the student write the correct idea?”)

✅ Matching systems (e.g. job descriptions vs resumes)

Code Using sentence-transformers for Sentence Similarity

We’ll use the model sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2
✅ It supports both English and Arabic!

```python
!pip install -U sentence-transformers
```

```python
from sentence_transformers import SentenceTransformer, util

# Load multilingual sentence similarity model
model = SentenceTransformer('sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2')

# Example sentence pairs
sentence1 = "Artificial intelligence will change the world."
sentence2 = "AI is going to transform our future."

# Encode both sentences
embedding1 = model.encode(sentence1, convert_to_tensor=True)
embedding2 = model.encode(sentence2, convert_to_tensor=True)

# Compute cosine similarity
similarity = util.cos_sim(embedding1, embedding2)

# Display result
print(f"Similarity Score: {similarity.item():.4f}")
```

# Task 8: Named Entity Recognition (NER)

What is NER?

Named Entity Recognition (NER) is the task of identifying and labeling
entities in a sentence, such as:

🧑 PERSON — names of people (e.g., Barack Obama)

🌍 LOCATION — cities, countries, etc. (e.g., Jordan, Paris)

🏢 ORGANIZATION — companies, groups (e.g., Google, UN)

🗓 DATE — years, time (e.g., 2023, July)

💲 MONEY — currency values (e.g., $500)

📏 QUANTITY, PERCENT, etc.



-----------------------------------------------

 Example:

Input text:

"Elon Musk founded SpaceX in 2002 and later became CEO of Tesla in California."


NER output:

Elon Musk → PERSON

SpaceX → ORGANIZATION

2002 → DATE

Tesla → ORGANIZATION

California → LOCATION



-------------------------------------------

Code using Hugging Face Pipeline
We'll use the model
dslim/bert-base-NER,
a strong general-purpose English NER model.

```python
from transformers import pipeline

# Load NER model
ner = pipeline("ner", model="dslim/bert-base-NER", grouped_entities=True)

# Input text
text = "Barack Obama was born in Hawaii and served as President of the United States from 2009 to 2017."

# Run NER
results = ner(text)

# Print results
print("Entities found:")
for r in results:
    print(f"{r['word']} → {r['entity_group']} (score: {r['score']:.2f})")
```

Code Explanation:

Line	What It Does

pipeline("ner")	Loads a pretrained NER model.

grouped_entities=True	Merges sub-word tokens into full names like "Barack Obama".

text = ...	Input your sentence here.

results = ner(text)	Run the model and get predictions.
Loop for r in results	Print out each entity + its type (PERSON, LOCATION, etc.)

```python
from transformers import pipeline

# Load the NER pipeline with grouped entities
ner = pipeline("ner", model="dslim/bert-base-NER", grouped_entities=True)

# Input text with many named entities
text = """
In 2023, Apple Inc. announced its new AI assistant during WWDC in California.
Meanwhile, Elon Musk was in Berlin discussing Tesla's expansion in Europe.
The United Nations held a climate summit in Geneva on October 14.
Microsoft and Google competed to acquire OpenAI, a San Francisco-based lab.
Cristiano Ronaldo signed a record-breaking contract with Al-Nassr FC in Saudi Arabia.
"""

# Run the NER model
results = ner(text)

# Print the results
print("Detected Entities:")
for r in results:
    print(f"{r['word']} → {r['entity_group']} (score: {r['score']:.2f})")
```
